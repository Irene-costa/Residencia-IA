```python
import os
import numpy as np
import pandas as pd
from sentence_transformers import SentenceTransformer

# 1. Carrega o modelo de embeddings
model = SentenceTransformer('sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2')

# 2. Função auxiliar para extrair embeddings
def get_embedding(texto: str) -> np.ndarray:
    return np.array(model.encode(texto), dtype=np.float32)

# 3. Funções de distância e similaridade solicitadas no exercício
def distancia_euclidiana(vec1: np.ndarray, vec2: np.ndarray) -> float:
    return float(np.linalg.norm(vec1 - vec2))

def similaridade_cosseno(vec1: np.ndarray, vec2: np.ndarray) -> float:
    norm_v1 = np.linalg.norm(vec1)
    norm_v2 = np.linalg.norm(vec2)
    if norm_v1 == 0 or norm_v2 == 0:
        return 0.0
    return float(np.dot(vec1, vec2) / (norm_v1 * norm_v2))

def distancia_cosseno(vec1: np.ndarray, vec2: np.ndarray) -> float:
    return float(1.0 - similaridade_cosseno(vec1, vec2))
```

```python
import numpy as np

def distancia_euclidiana(embedding_a, embedding_b):
    """Calcula a distância euclidiana entre dois vetores."""
    a = np.array(embedding_a)
    b = np.array(embedding_b)
    return np.linalg.norm(a - b)

def distancia_cosseno(embedding_a, embedding_b):
    """Calcula a distância de cosseno (1 - similaridade de cosseno) entre dois vetores."""
    a = np.array(embedding_a)
    b = np.array(embedding_b)

    dot_product = np.dot(a, b)
    norm_a = np.linalg.norm(a)
    norm_b = np.linalg.norm(b)

    if norm_a == 0 or norm_b == 0:
        return 1.0

    similaridade = dot_product / (norm_a * norm_b)
    return 1 - similaridade
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.manifold import MDS

# 1. Palavras da atividade e extração dos vetores
palavras = ["gato", "felino", "cachorro", "carro", "caminhao", "moto", "banana", "maça", "goiaba"]
vetores = np.array([model.encode(p) for p in palavras])

# 2. Reduz os embeddings (384D) para 2 dimensões usando MDS
mds = MDS(n_components=2, dissimilarity='euclidean', random_state=42)
coords = mds.fit_transform(vetores)

# 3. Desenha o gráfico
plt.figure(figsize=(9, 7))

for i, palavra in enumerate(palavras):
    x, y = coords[i]
    plt.scatter(x, y, s=120, label=palavra)
    plt.annotate(palavra, (x, y), textcoords="offset points", xytext=(0, 10), ha='center', fontsize=11)

plt.title("Proximidade de Termos Baseada na Similaridade (MDS)", fontsize=13)
plt.xlabel("Componente 1 (MDS)", fontsize=11)
plt.ylabel("Componente 2 (MDS)", fontsize=11)
plt.legend(bbox_to_anchor=(1.02, 1), loc='upper left')

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.manifold import MDS

# 1. Palavras e extração dos vetores
palavras = ["gato", "felino", "cachorro", "carro", "caminhao", "moto", "banana", "maça", "goiaba"]
vetores = np.array([model.encode(p) for p in palavras])

# 2. Reduz os embeddings para 3 dimensões (n_components=3)
mds = MDS(n_components=3, dissimilarity='euclidean', random_state=42)
coords_3d = mds.fit_transform(vetores)

# 3. Cria a figura em 3D
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

# 4. Plota cada ponto no espaço tridimensional (X, Y, Z)
for i, palavra in enumerate(palavras):
    x, y, z = coords_3d[i]
    ax.scatter(x, y, z, s=120, label=palavra)
    ax.text(x, y, z + 0.1, palavra, fontsize=10)

ax.set_title("Proximidade de Termos Baseada na Similaridade em 3D (MDS)", fontsize=12)
ax.set_xlabel("Componente 1 (MDS)")
ax.set_ylabel("Componente 2 (MDS)")
ax.set_zlabel("Componente 3 (MDS)")
ax.legend(bbox_to_anchor=(1.15, 1), loc='upper left')

plt.tight_layout()
plt.show()
```
